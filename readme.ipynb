# Windows focused cybersecurity threat knowledge graph

This project has made a cybersecurity threat knowledge knowledge graph for Windows. It integrated ATT&CK and CVE data through RDF triple in a knowledge graph.

## Requirements
Download Python <br>
Jupyter Notebook or Visual Studio Code<br>
rdflib<br>
pandas<br>
matplotlib<br>
### How to reproduce the project
####OPtion 1
1. Clone repository
2. Install required packages
3. Run analysis.ipynb. This notebook loads windows-cyber-threat-ontology.ttl and windows-cyber-threat-instances.ttl from Github. Then it will run the SPARSQL get the information and creates graph base on data

####Option2
1. Clone repository
2. Open importdata.ipynb.
3. Run all cells except the final save cell.
4. In the final cell, set save_place to the location where the generated instance file should be saved
5. Run the final cell
6. Open analysis.ipynb
7. Replace its GitHub URLs with the paths of the local TTL files
8. Run the file

For example, on Windows:
"C:\Users\YourName\Documents\windows-cyber-threat-instances.ttl"
To save the file in the same directory as the notebook, use:
save_place = "windows-cyber-threat-instances.ttl"
g.serialize(destination=save_place, format="turtle")
